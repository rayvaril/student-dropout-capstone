# Early Warning for Student Attrition
## An Explainable and Fair Machine Learning Approach

### 00 — Problem Definition

This notebook defines the capstone problem before any data analysis or modeling is performed.

The goal is to establish:
- the business problem
- the machine learning task
- the target outcome
- the meaning of "early warning"
- technical success metrics
- operational or business success metrics
- fairness objectives
- key assumptions and scope


## 1. Background and Context

Student attrition is a significant challenge for higher-education institutions because students who leave before completing their studies may experience academic, financial, and personal consequences, while institutions also face losses in retention and completion outcomes.

Intervention is often most useful when institutions can identify risk **before** a student's situation becomes difficult to reverse.

This project explores whether machine learning can support earlier identification of students at risk of dropping out, while keeping the predictions explainable and evaluating whether model performance is comparable across student groups.


## 2. Business Problem

Higher-education institutions have limited resources for proactive student support.

Counselors, academic advisers, and student-support teams may not be able to intervene with every student at the same intensity. A useful early-warning system should therefore help institutions identify students who may benefit from additional support while avoiding over-reliance on opaque or potentially biased predictions.

### Business Problem Statement

**How can an institution identify students at elevated risk of dropping out early enough to support timely and targeted intervention, while using limited support resources responsibly?**


## 3. Research Question

### Primary Research Question

**Can machine learning identify students at risk of dropping out early enough to support targeted intervention, while maintaining explainable and fair predictive performance across student groups?**

### Supporting Questions

1. How accurately can student dropout risk be predicted using information available at different stages of the student journey?
2. Which factors contribute most strongly to predicted dropout risk?
3. How does predictive performance change when later academic information is added?
4. Does the model perform differently across relevant student groups?
5. If intervention capacity is limited, how effectively can the model prioritize students who may need support?


## 4. Machine Learning Task

The primary machine learning task will be **supervised classification**.

The original dataset contains three outcomes:

- `Dropout`
- `Enrolled`
- `Graduate`

For the primary analysis, the target will initially be simplified into a binary outcome:

- `1` = Dropout
- `0` = Non-Dropout (`Enrolled` or `Graduate`)

The binary framing keeps the project focused on the early identification of dropout risk.

A multiclass extension may be explored later only if it adds meaningful value.


## 5. Defining "Early Warning"

The project will distinguish between information available at different stages.

### Model A — Enrollment-Stage Early Warning

Uses only information available at or near enrollment.

Purpose:
- maximize intervention lead time
- test how much dropout risk can be identified before academic performance is observed

### Model B — First-Semester Early Warning

Adds first-semester academic information.

Purpose:
- test whether early academic performance materially improves prediction
- quantify the trade-off between **earlier intervention** and **stronger predictive information**

Second-semester variables will not be used in the primary early-warning models because they may arrive too late to support the intended intervention use case.


## 6. Technical Success Metrics

The project will not rely on accuracy alone.

Because the most costly modeling error is failing to identify a student who eventually drops out, the primary metric will be:

### Primary Metric
**Recall for the Dropout class**

This measures how many actual dropout cases the model successfully identifies.

### Supporting Metrics

- Precision
- F1-score
- ROC-AUC
- PR-AUC
- Confusion Matrix

These metrics will be evaluated across candidate models before selecting the final model.


## 7. Operational Success Metric

The model should also be evaluated in a way that reflects limited support capacity.

### Intervention Scenario

Assume the institution can proactively intervene with only the **top 10% of students ranked by predicted dropout risk**.

A key operational metric will therefore be:

### Recall@Top 10%

**Of all students who eventually drop out, what proportion would be identified if the institution could support only the 10% of students with the highest predicted risk?**

This converts model performance into a practical decision-support scenario.


## 8. Explainability Objective

The system should not only identify high-risk students but also provide insight into the factors driving predictions.

Explainability analysis will be used to examine:

- overall feature importance
- the strongest drivers of dropout risk
- individual prediction explanations
- the difference between predictive and potentially actionable factors

The model is intended to support human decision-making, not replace academic advisers or automatically determine student outcomes.


## 9. Fairness Objective

The project will evaluate whether model performance differs meaningfully across student groups.

Potential fairness dimensions include:

- gender
- age groups
- scholarship status
- socioeconomic indicators, where appropriate

Metrics may include:

- recall by group
- false-negative rate by group
- selection rate
- demographic parity difference
- equalized odds difference

Fairness mitigation will be considered if meaningful disparities are observed.


## 10. Key Assumptions

1. The dataset is representative enough to demonstrate the machine learning lifecycle, but results should not automatically be generalized to all universities or student populations.
2. A prediction of high dropout risk does not imply that a student will definitely drop out.
3. Sensitive or demographic variables require careful interpretation.
4. The system should be used to **offer support**, not restrict opportunities or penalize students.
5. Feature availability must reflect when information would realistically be known in an operational setting.


## 11. Project Scope

### In Scope

- binary dropout-risk classification
- comparison of multiple machine learning models
- enrollment-stage and first-semester prediction
- explainability analysis
- fairness evaluation
- limited-capacity intervention prioritization
- technical and business communication

### Out of Scope for the Core Project

- automated student intervention
- causal claims about why students drop out
- clinical or psychological profiling
- production-scale deployment
- generative AI features

These may be considered only as optional extensions after the core capstone requirements are completed.


## 12. Expected Project Output

The completed capstone should answer four practical questions:

### Predict
Who appears to be at elevated risk of dropping out?

### Explain
Which factors are contributing to that prediction?

### Evaluate Fairness
Does the model perform comparably across relevant student groups?

### Support Decisions
If intervention resources are limited, which students should be prioritized for proactive support?
